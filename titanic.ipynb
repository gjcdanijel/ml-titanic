```python
# Assume that Parents/Childs Siblings/Spouses share same cabin, group them with there surname

df['Surname'] = df['Name'].str.split(',').str[0].str.strip()

def find_cabin(row):
    if pd.notna(row['Cabin']):
        return row['Cabin']
    if row['SibSp'] == 0 and row['Parch'] == 0:
        return row['Cabin']

    family = df[
        (df['Surname'] == row['Surname'])
        & (df.index != row.name)
        & df['Cabin'].notna()
    ]

    if row['Parch'] > 0:
        parch = family[family['Parch'] > 0]
        if not parch.empty:
            return parch['Cabin'].iloc[0]

    if row['SibSp'] > 0:
        sibsp = family[family['SibSp'] > 0]
        if not sibsp.empty:
            return sibsp['Cabin'].iloc[0]

    return row['Cabin']

print("Missing before:", df['Cabin'].isna().sum())
df['Cabin'] = df.apply(find_cabin, axis=1)
print("Missing after:", df['Cabin'].isna().sum())
```

```python
cabins = df[['Pclass', 'Cabin']].dropna().copy()

# split multi-cabin entries like "C23 C25 C27" into separate rows
cabins['Cabin'] = cabins['Cabin'].str.split()
cabins = cabins.explode('Cabin')

# remove the numbers, keep only the letters
cabins['Deck'] = cabins['Cabin'].str.replace(r'\d', '', regex=True)

# link each class to its sorted list of unique letters
class_decks = cabins.groupby('Pclass')['Deck'].apply(lambda s: sorted(s.unique()))

for pclass, letters in class_decks.items():
    print(f"Class {pclass}: {', '.join(letters)}")
```

```python
df['Deck'] = df['Cabin'].str[0]
df['HasCabin'] = df['Cabin'].notna().astype(int)  # keep the real info before filling

def balance_decks(df, exclude=('T',)):
    df = df.copy()
    for pclass, group in df.groupby('Pclass'):
        known = group['Deck'].dropna()
        counts = known[~known.isin(exclude)].value_counts()
        counts = dict(sorted(counts.items()))
        if not counts:
            continue

        missing_idx = group.index[group['Deck'].isna()]
        print(f"Class {pclass}")
        print(f"  Before: {counts}  | missing: {len(missing_idx)}")

        assigned = []
        for _ in missing_idx:
            deck = min(counts, key=counts.get)
            counts[deck] += 1
            assigned.append(deck)

        df.loc[missing_idx, 'Deck'] = assigned
        # print(f"  After:  {counts}\n")
    return df

df = balance_decks(df)
```

```python
df['Deck'] = df['Cabin'].str[0]
df['HasCabin'] = df['Cabin'].notna().astype(int)  # keep the real info before filling

def balance_decks(df, exclude=('T',)):
    df = df.copy()
    for pclass, group in df.groupby('Pclass'):
        known = group['Deck'].dropna()
        counts = known[~known.isin(exclude)].value_counts()
        counts = dict(sorted(counts.items()))  # alphabetical, for stable tie-breaking
        if not counts:
            continue

        missing_idx = group.index[group['Deck'].isna()]
        print(f"Class {pclass}")
        print(f"  Before: {counts}  | missing: {len(missing_idx)}")

        assigned = []
        for _ in missing_idx:
            deck = min(counts, key=counts.get)  # deck with the fewest passengers
            counts[deck] += 1
            assigned.append(deck)

        df.loc[missing_idx, 'Deck'] = assigned
        print(f"  After:  {counts}\n")
    return df
```

```python
df['Deck'] = df['Cabin'].str[0]
df['HasCabin'] = df['Cabin'].notna().astype(int)
df['CabinImputed'] = df['Cabin'].isna().astype(int)

df = balance_decks(df)

taken = set(df['Cabin'].dropna().str.split().explode())

def next_free_cabin(deck, taken):
    n = 1
    while f"{deck}{n}" in taken:
        n += 1
    return f"{deck}{n}"

missing = df['Cabin'].isna() & df['Deck'].notna()
for idx in df.loc[missing].sort_values('Pclass').index:
    cabin = next_free_cabin(df.at[idx, 'Deck'], taken)
    taken.add(cabin)
    df.at[idx, 'Cabin'] = cabin

print("Missing cabins left:", df['Cabin'].isna().sum())
```
